### DATA INGESTION


```python
###Document structure


from langchain_core.documents import Document
```

```python
###Document creation

doc = Document(
    page_content="This is the content of the document.",
    metadata={"source": "example.txt",
    "pages": 1,
    "author": "John Doe",
    "date": "2022-01-01"},
)
print(doc)  # This will print the document structure
```

```python
### CREATE A TEXT FILE

import os
os.makedirs("data/text_files", exist_ok=True)
```

```python
sample_text = """This is a sample text file.
It contains multiple lines of text.
"""  
with open("data/text_files/sample.txt", "w") as f:
    f.write(sample_text)
print("Text file created successfully.")
```

```python
###Text Loader or reader

from langchain_community.document_loaders import TextLoader
loader = TextLoader("data/text_files/sample.txt")
documents = loader.load()
print(documents)
```

```python
###Directory Loader

from langchain_community.document_loaders import DirectoryLoader, TextLoader
loader = DirectoryLoader("data/text_files/", glob="*.txt", loader_cls=TextLoader)
documents = loader.load()
print(documents)
```

```python
### PDF Loader

from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("data/pdf/sample.pdf")
documents = loader.load()
print(documents)
```
